### Need to combine movies and links dataset and use tmdbId to get movie info with api and add more details to the dataset

```python
import requests
from tqdm import tqdm
import time
from concurrent.futures import ThreadPoolExecutor, as_completed
# import logging
```

```python
API_KEY = "e0dd5e1df411ea273a7b3bdfd8d3df78"

def get_movie_details(tmdb_id, max_retries=3):
   
    url = f"https://api.themoviedb.org/3/movie/{tmdb_id}"
    params = {
        "api_key": API_KEY,
        "append_to_response": "credits"
    }
    
    # Try twice
    for attempt in range(max_retries):
        try:
            response = requests.get(url, params=params, timeout=5)
            
            if response.status_code == 200:
                data = response.json()
                
                overview = data.get("overview", "")
                
                # Get Director
                director = ""
                for crew in data.get("credits", {}).get("crew", []):
                    if crew.get("job") == "Director":
                        director = crew.get("name", "")
                        break
                
                # Get Top 5 Actors
                cast = data.get("credits", {}).get("cast", [])
                actors = ", ".join([actor.get("name", "") for actor in cast[:5] if actor.get("name")])
                
                return overview, director, actors
            
            elif response.status_code == 404:
                return "", "", ""  # Movie doesn't exist
            
        except:
            if attempt == 0:  # First attempt failed, try once more
                time.sleep(0.5)
                continue
    
    return np.nan, np.nan, np.nan  # Both attempts failed


def get_movie_details_with_id(tmdb_id, max_retries=3):
    """Wrapper that returns tmdb_id with results"""
    overview, director, actors = get_movie_details(tmdb_id, max_retries)
    return tmdb_id, overview, director, actors

def super_fast_enrich_refetch(df, ind, num_threads=15, max_retries=3):
    
    results = []
    
    with ThreadPoolExecutor(max_workers=num_threads) as executor:
        futures = [executor.submit(get_movie_details_with_id, tmdb_id, max_retries) 
                  for tmdb_id in df.iloc[ind]["tmdbId"].tolist()]
        
        for future in tqdm(as_completed(futures), total=len(futures), desc="Fetching"):
            results.append(future.result())
    
    #display(results)    
    cols = ['tmdbId', 'overview', 'director', 'actors']
    df.loc[df.index[ind], cols] = results
    
    return df


def super_fast_enrich(df, num_threads=15, max_retries=3):
    
    results = []
    
    with ThreadPoolExecutor(max_workers=num_threads) as executor:
        futures = [executor.submit(get_movie_details_with_id, tmdb_id, max_retries) 
                  for tmdb_id in df["tmdbId"].tolist()]
        
        for future in tqdm(as_completed(futures), total=len(futures), desc="Fetching"):
            results.append(future.result())
    
    # Create results DataFrame
    results_df = pd.DataFrame(results, columns=['tmdbId', 'overview', 'director', 'actors'])
    
    # Drop old columns if they exist
    df = df.drop(columns=['overview', 'director', 'actors'], errors='ignore')
    
    # Merge
    df = df.merge(results_df, on='tmdbId', how='left')
    
    return df


# STEP 1: Load your data
# df = movies_meta.copy()

# # STEP 2: Enrich with retry logic (10 attempts per movie)
# df = super_fast_enrich(df, num_threads=20, max_retries=7)

# # Refetching if any missing vals
# ind = df[df['actors'].isna()].index.to_list()
# df = super_fast_enrich_refetch(df, ind, 20, 5)

# df
```
